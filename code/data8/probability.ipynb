### 1. When an Event does not happen

P(an event doesn't happen) = 1 - P(the event happens)

### 2. When all outcomes are equally likely

P(an event happens) = #{outcomes that make the event happen} / #{all outcomes}

### 3. When two events must both happen
Multiplication Rule

P(two events both happen) = P(one event happens) x P(the other event happens, given that the first one happened)

### 4. When an event can happen in two different ways
Addition Rule

P(an event happens) = P(first way it can happen) + P(the second way it can happen)
