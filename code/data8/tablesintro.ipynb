## Rows again with `where`

### Multiple Features and `where`

```python
nba.where("POSITION", "PG").where("SALARY", are.above(15))
```

```python
nba.where("SALARY", are.between(10, 10.5))
```

```python
nba.where("TEAM", are.containing("Warriors"))
```

```python
# get both PG and SG (point guards and shooting guards)
nba.where("POSITION", are.containing("G"))
```

```python
nba.where("TEAM", are.not_equal_to("Cleveland Cavaliers")).where("SALARY", are.not_below(20))
```
