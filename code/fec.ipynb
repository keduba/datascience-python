### Donation statistics by occupation

```python
fec["contbr_occupation"].value_counts().head(20)
```

Since some occupations listed are merely repeats in different words, we'll clean them up.

```python
occ_mapping = {
   "INFORMATION REQUESTED PER BEST EFFORTS" : "NOT PROVIDED",
   "INFORMATION REQUESTED" : "NOT PROVIDED",
   "INFORMATION REQUESTED (BEST EFFORTS)" : "NOT PROVIDED",
   "C.E.O.": "CEO",
   "OWNER" : "SELF-EMPLOYED"
}

def get_occ(x):
    # If no mapping provided, return x
    return occ_mapping.get(x, x)

fec.loc[:, "contbr_occupation"] = fec["contbr_occupation"].map(get_occ)
```

```python
emp_mapping = {
   "INFORMATION REQUESTED PER BEST EFFORTS" : "NOT PROVIDED",
   "INFORMATION REQUESTED" : "NOT PROVIDED",
   "SELF" : "SELF-EMPLOYED",
   "SELF EMPLOYED" : "SELF-EMPLOYED",
   "OWNER" : "SELF-EMPLOYED"
}

def get_emp(x):
    # If no mapping provided, return x
    return emp_mapping.get(x, x)

fec.loc[:, "contbr_employer"] = fec["contbr_employer"].map(get_emp)
```

```python
# Aggregate by party and occupation. Hint: pivot table
ocupado = fec.pivot_table("contb_receipt_amt", index="contbr_occupation", columns="party", aggfunc="sum")
ocupado
```

```python
over2M = ocupado[ocupado.sum(axis=1) > 2_000_000]
over2M
```

```python
over2M.plot(kind="barh", title="Donations over $2M")
```

### Top Donor Occupations to Obama and Romney

```python
def top_amounts(group, key, n):
    totals = group.groupby(key)["contb_receipt_amt"].sum()
    return totals.nlargest(n)
    
```

```python
fec_bomr = fec[fec["cand_nm"].isin(["Obama, Barack", "Romney, Mitt"])].reset_index(drop=True)
```

```python
grp_mo = fec_bomr.groupby("cand_nm")
grp_mo.apply(top_amounts, "contbr_occupation", 7, include_groups=False)
```

```python
grp_mo.apply(top_amounts, "contbr_employer", 10, include_groups=False)
```

### Donation amounts by states

```python
groupedmo = fec_bomr.groupby(["cand_nm", "contbr_st"])

totalsmo = groupedmo["contb_receipt_amt"].sum().unstack(level=0).fillna(0)
```

```python
totalsmo
```

```python
totalsmo = totalsmo[totalsmo.sum(axis="columns") > 100_000]
totalsmo
```

```python
# Relative percentage contribution per state

percent = totalsmo.div(totalsmo.sum(axis="columns"), axis="index")
percent
```
